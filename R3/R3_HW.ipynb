```python
from langchain_huggingface import HuggingFacePipeline
from langchain_community.tools import Tool
from langchain.agents import initialize_agent, AgentType
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

model_name = "Qwen/Qwen3-1.7B"
tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(model_name, trust_remote_code=True)

# 建立推理 pipeline
pipe = pipeline("text-generation", model=model, tokenizer=tokenizer, max_new_tokens=256, device=0)

# 包裝成 LangChain LLM
llm = HuggingFacePipeline(pipeline=pipe)

# 定義三個工具
def get_weather(location: str) -> str:
    return f"The weather in {location} is sunny with 25°C."

def get_time(location: str) -> str:
    return f"The current time in {location} is 14:00."

def get_exchange_rate(currency_pair: str) -> str:
    return f"The exchange rate for {currency_pair} is 1.1."

tools = [
    Tool(
        name="Weather Checker",
        func=get_weather,
        description="Use this tool to check the weather in a specific location."
    ),
    Tool(
        name="Time Checker",
        func=get_time,
        description="Use this tool to check the current time in a specific location."
    ),
    Tool(
        name="Exchange Rate Checker",
        func=get_exchange_rate,
        description="Use this tool to check the exchange rate between two currencies, format: USD/EUR."
    )
]

# 初始化 Agent
agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

# 測試
agent.run("What's the weather in Tokyo?")
agent.run("What time is it in New York?")
agent.run("What's the exchange rate for USD/JPY?")
```
